# Dataset Exploration & Sanity Check

Quick notebook to verify the downloaded dataset looks correct.

```python
from pathlib import Path
import yaml
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python
# TODO: Update this path to match your downloaded dataset folder
DATA_ROOT = Path(".")  # e.g. Path("URC-2024-...")

# Load data.yaml
with open(DATA_ROOT / "data.yaml") as f:
    data_cfg = yaml.safe_load(f)

print("Classes:", data_cfg.get("names", []))
print("Num classes:", data_cfg.get("nc", "?"))
```

```python
# Count images per split
for split in ["train", "valid", "test"]:
    img_dir = DATA_ROOT / split / "images"
    if img_dir.exists():
        count = len(list(img_dir.iterdir()))
        print(f"{split}: {count} images")
    else:
        print(f"{split}: not found")
```

```python
# Show a few sample images
train_imgs = sorted((DATA_ROOT / "train" / "images").glob("*"))
fig, axes = plt.subplots(1, min(4, len(train_imgs)), figsize=(16, 4))
if not isinstance(axes, list):
    axes = [axes]
for ax, img_path in zip(axes, train_imgs[:4]):
    ax.imshow(Image.open(img_path))
    ax.set_title(img_path.name, fontsize=8)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
